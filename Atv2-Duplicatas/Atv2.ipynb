```python
# IMPORTAÇÕES
import pandas as pd
import numpy as np
```

```python
# CARREGAMENTO DOS DADOS
arquivo = 'KPMG_VI_New_raw_data_update_final.xlsx'

df = pd.read_excel(arquivo, sheet_name='Transactions', header=1)

print(df.head())
```

```python
# CONFERÊNCIA DO CABEÇALHO E DOS TIPOS

print("\nNome das colunas:")
print(df.columns)

print("\nTipos de dados:")
print(df.dtypes)

print("\nQuantidade de linhas e colunas:")
print(df.shape)
```

```python
# DUPLICATAS EXATAS

# Verifica se existem linhas iguais
print("\nDuplicatas exatas:")
print(df.duplicated().sum())

# Mostra as linhas repetidas
print(df[df.duplicated(keep=False)])

```

```python
# DUPLICATAS POR SUBCONJUNTO DE COLUNAS

# Verifica transações com o mesmo cliente, produto,
# data e preço
print("\nDuplicatas por algumas colunas:")

print(df.duplicated(subset=[
    'customer_id',
    'product_id',
    'transaction_date',
    'list_price'
]).sum())
```

```python
# UNICIDADE DOS IDENTIFICADORES

# Verifica se os IDs são únicos
print("\nOs IDs são únicos?")
print(df['transaction_id'].is_unique)

# Conta quantos IDs estão repetidos
print("\nIDs repetidos:")
print(df['transaction_id'].duplicated().sum())

# Verifica se existem IDs vazios
print("\nIDs vazios:")
print(df['transaction_id'].isnull().sum())
```

```python
# ESTRATÉGIA DE TRATAMENTO E IMPACTO

# Faz uma cópia dos dados originais
df_tratado = df.copy()

# Remove as linhas completamente iguais
df_tratado = df_tratado.drop_duplicates()

# Compara a quantidade de linhas
print("\nQuantidade de linhas:")

print("Antes:", len(df))
print("Depois:", len(df_tratado))

# Calcula quantas linhas foram removidas
removidos = len(df) - len(df_tratado)

print("\nLinhas removidas:")
print(removidos)


# Calcula a soma dos preços antes e depois
preco_antes = df['list_price'].sum()

preco_depois = df_tratado['list_price'].sum()

print("\nSoma dos preços antes:")
print(preco_antes)

print("\nSoma dos preços depois:")
print(preco_depois)

# Calcula a diferença dos preços
diferenca = preco_antes - preco_depois

print("\nDiferença dos preços:")
print(diferenca)
```
