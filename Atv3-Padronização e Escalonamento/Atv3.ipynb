```python
# IMPORTAÇÕES

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

```python
# NOMES DAS COLUNAS (consulte a documentação do dataset)
colunas = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]
```

```python
# CARREGAMENTO DOS DADOS
df = pd.read_csv('adult/adult.data', header=None, names=colunas)

print(df.head())

print("\nQuantidade de linhas e colunas:")
print(df.shape)
```

```python
# INSPEÇÃO DAS COLUNAS CATEGÓRICAS

# Pegando as colunas de texto
colunas_texto = df.select_dtypes(include='object').columns

# Mostrando os valores de cada coluna
for coluna in colunas_texto:
    print("\n", coluna)
    print(df[coluna].unique())
```

```python
# PADRONIZAÇÃO DOS TEXTOS (ANTES x DEPOIS)

# Fazendo uma cópia dos dados
df_limpo = df.copy()

# Quantidade de categorias antes
print("\nCategorias antes:")
print(df[colunas_texto].nunique())

# Tirando os espaços extras
for coluna in colunas_texto:
    df_limpo[coluna] = df_limpo[coluna].str.strip()

# Quantidade de categorias depois
print("\nCategorias depois:")
print(df_limpo[colunas_texto].nunique())
```

```python

# CONVERSÃO DOS MARCADORES DE AUSÊNCIA

# Valores ausentes antes
antes = df_limpo.isnull().sum()

# Trocando ? por NaN
df_limpo = df_limpo.replace('?', np.nan)

# Valores ausentes depois
depois = df_limpo.isnull().sum()

print("\nValores ausentes antes:")
print(antes)

print("\nValores ausentes depois:")
print(depois)

print("\nNovos valores ausentes:")
print(depois - antes)
```

```python
# SELEÇÃO DAS COLUNAS NUMÉRICAS

# Pegando apenas as colunas numéricas
numericas = df_limpo.select_dtypes(include='number').columns

print("\nColunas numéricas:")
print(numericas)
```

```python
# ESCALONAMENTO — MIN-MAX

# Fazendo uma cópia
df_minmax = df_limpo.copy()

# Criando o escalonador
minmax = MinMaxScaler()

# Aplicando o Min-Max
df_minmax[numericas] = minmax.fit_transform(df_minmax[numericas])

print("\nMin-Max:")
print(df_minmax[numericas].head())
```

```python
# ESCALONAMENTO — Z-SCORE

# Fazendo uma cópia
df_zscore = df_limpo.copy()

# Criando o escalonador
zscore = StandardScaler()

# Aplicando o Z-score
df_zscore[numericas] = zscore.fit_transform(df_zscore[numericas])

print("\nZ-score:")
print(df_zscore[numericas].head())
```

```python
# ESCALONAMENTO — ROBUSTO

# Fazendo uma cópia
df_robusto = df_limpo.copy()

# Criando o escalonador
robusto = RobustScaler()

# Aplicando o escalonamento robusto
df_robusto[numericas] = robusto.fit_transform(df_robusto[numericas])

print("\nRobusto:")
print(df_robusto[numericas].head())
```

```python
# COMPARAÇÃO ENTRE AS TÉCNICAS

# Comparando mínimo e máximo
print("\nMínimo e máximo - Min-Max:")
print(df_minmax[numericas].agg(['min', 'max']))

print("\nMínimo e máximo - Z-score:")
print(df_zscore[numericas].agg(['min', 'max']))

print("\nMínimo e máximo - Robusto:")
print(df_robusto[numericas].agg(['min', 'max']))


# Comparando as médias
print("\nMédia - Min-Max:")
print(df_minmax[numericas].mean())

print("\nMédia - Z-score:")
print(df_zscore[numericas].mean())

print("\nMédia - Robusto:")
print(df_robusto[numericas].mean())


# Comparando o desvio-padrão
print("\nDesvio-padrão - Min-Max:")
print(df_minmax[numericas].std())

print("\nDesvio-padrão - Z-score:")
print(df_zscore[numericas].std())

print("\nDesvio-padrão - Robusto:")
print(df_robusto[numericas].std())


# Comparando uma coluna com valores extremos
print("\nCapital-gain original:")
print(df_limpo['capital-gain'].describe())

print("\nCapital-gain Min-Max:")
print(df_minmax['capital-gain'].describe())

print("\nCapital-gain Z-score:")
print(df_zscore['capital-gain'].describe())

print("\nCapital-gain Robusto:")
print(df_robusto['capital-gain'].describe())
```
